## 2. 创建 DataFrame

> pandas 的核心是 **DataFrame**（表格）和 **Series**（一列）。它们可以存储数据，并对这些数据进行方便快捷的操作计算  
>DataFrame的创建有很多方法，从字典创建最为直观：key 是列名，value 是列数据。

```python
# 用字典创建一个小 DataFrame
df = pd.DataFrame({
    'name':    ['Alice', 'Bob',   'Charlie', 'David', 'Eve'],
    'age':     [25,      30,      35,        28,      30],
    'salary':  [5000,    6000,    7000,      5500,    6000],
    'dept':    ['IT',    'HR',    'IT',      'HR',    'IT']
})
df
```

```python
# Series：DataFrame 的一列
print(df['name'])       # 取出 name 列
print(type(df['name']))
```

## 5. 计算统计量

> 对数值列算均值、中位数等，快速了解数据的集中趋势和离散程度。

```python
print('平均工资:', df['salary'].mean())
print('工资中位数:', df['salary'].median())
print('标准差:',   df['salary'].std())
print('最低工资:',  df['salary'].min())
print('最高工资:',  df['salary'].max())
```

## 7. 缺失值

> 三步走：**检测 -> 填充 -> 删除**。  

```python
# 创建一个带缺失值的 DataFrame
df_miss = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': [1, 2, 3, 4, 5]
})
print('原始数据:')
print(df_miss)
```

```python
# 检测
print('逐列缺失数:')
print(df_miss.isnull().sum())
```

```python
# 填充
df_filled = df_miss.copy()
df_filled['A'] = df_filled['A'].fillna(df_filled['A'].mean())  # 用均值
df_filled['B'] = df_filled['B'].fillna(0)                       # 用固定值，如0
print('填充后:')
print(df_filled)
```

```python
# 删除含 NaN 的行
print('删除含 NaN 的行:')
print(df_miss.dropna())
```

## 8. 分组聚合

> `groupby` 先按某列的值把数据拆成若干组，再对每组算统计值。
>
> 比如我们按 `dept`（部门）分组，数据会被拆成两组：
> - **IT 组** → Alice、Charlie、Eve（3 人）
> - **HR 组** → Bob、David（2 人）
>
> 拆完后就能算"IT 平均工资多少？HR 平均工资多少？"——这就是聚合。

```python
print('各部门平均工资:')
print(df.groupby('dept')['salary'].mean())
print()
print('各部门人数:')
print(df['dept'].value_counts())
```

```python
# 一次算多个统计量
df.groupby('dept')['salary'].agg(['mean', 'max', 'min', 'count'])
```

## 9. 新增列

> 基于已有列创建新列。  

```python
df['bonus'] = df['salary'] * 0.2          # 向量化运算
df['total'] = df['salary'] + df['bonus']
df[['name', 'salary', 'bonus', 'total']]
```

## 10. 排序

> `.sort_values('要排序的列名', ascending=True)`——默认升序（小→大）。  
> `ascending=False` 即降序（大→小）。  
> 多列排序时传列表，每列可分别指定升/降序。

```python
print('按工资降序（高→低）:')
print(df.sort_values('salary', ascending=False)[['name', 'salary']])
print()
print('按工资升序（低→高，默认就是升序）:')
print(df.sort_values('salary')[['name', 'salary']])
print()
print('先按部门、再按工资降序:')
print(df.sort_values(['dept', 'salary'], ascending=[True, False]))
```

## 11. 合并

> `concat` 纵向堆叠，`merge` 横向拼接。  

```python
# concat：按行堆叠
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
print('纵向堆叠:')
print(pd.concat([df1, df2]))
print()
# merge：按 key 拼接
left  = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
right = pd.DataFrame({'id': [1, 2, 4], 'score': [85, 92, 78]})
print('按 id 合并(inner):')
print(pd.merge(left, right, on='id', how='inner'))
```
